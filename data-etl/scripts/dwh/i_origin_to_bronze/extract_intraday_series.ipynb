# Intraday extractions

## Introduction 
This notebook will extract data from Alpha Vantage api intraday (stocks market 
values).  
After extracting data, the files will be saved in ... format in MinIo storage, 
inside bronze layer. 

## Utils

### Libraries & Paths

```python
import requests 
import os 

APIKEY = os.getenv('ALPHA_VANTAGE_API_KEY')
```

```python
base_url = 'https://www.alphavantage.co/'
endpoint = 'query'
params = {
    'function': 'TIME_SERIES_INTRADAY', # TODO, this should be parameterized to extract different stocks functions with the same code
    'symbol': 'AAPL', # TODO, this should be parameterized to extract different stocks values with the same code
    'interval': '5min', # TODO, this should be parameterized to extract different stocks frequency with the same code
    'apikey': APIKEY
}

response = requests.get(base_url + endpoint, params=params)
```
